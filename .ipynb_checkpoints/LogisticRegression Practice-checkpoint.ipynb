```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import math
```

```python
dataset=pd.read_csv(r"C:\Users\HP\Desktop\PG\ML\student_data_raw_2000.csv")
```

```python
dataset.isnull().sum()
```

```python
dataset.dropna(subset=['Placed_Tech'],inplace=True)
```

```python
dataset
```

```python
dataset['City'].unique()
```

```python
def replace(string):
    if(string=='NEW DELHI' or string=='New  Delhi'):
        return 'New Delhi'
    elif(string=='ghaziabad' or string=='Gaziabad'):
        return 'Ghaziabad'
    elif(string=='gorakhpur'):
        return 'Gorakhpur'
    elif(string=='Roorkee '):
        return 'Roorkee'
    else:
        return string
dataset['City']=(dataset['City'].apply(replace))
```

```python
dataset
```

```python
dataset['City'].unique()
```

```python
dataset['Deep_Learning_Hours']=(dataset['Deep_Learning_Hours'].apply(lambda n: abs(n)))
dataset
```

```python
dataset.groupby(['City'])['StudentID'].count()
```

```python
dataset.isnull().sum()
```

```python
dataset['C++_Score'].fillna(dataset['C++_Score'].median(),inplace=True)
dataset['Deep_Learning_Hours'].fillna(dataset['Deep_Learning_Hours'].median(),inplace=True)
dataset['Cloud_Hackathon'] = dataset.groupby('Placed_Tech')['Cloud_Hackathon'].transform(lambda x: x.fillna(x.mode()[0]))


dataset['Cloud_Hackathon'] = dataset['Cloud_Hackathon'].str.lower().map({
    'yes': 1, 'y': 1,
    'no': 0, 'n': 0
})
```

```python
dataset['Cloud_Hackathon'].unique()
```

```python
X=dataset[['C++_Score','Deep_Learning_Hours','Cloud_Hackathon']]
X
```

```python
y=dataset['Placed_Tech']
y
```

```python
%matplotlib inline
plt.figure(figsize=(12,8))
sns.scatterplot(x='C++_Score',y='Deep_Learning_Hours',data=dataset,hue='Placed_Tech');
```

```python
sns.scatterplot(x='C++_Score',y='Cloud_Hackathon',data=dataset,hue='Placed_Tech');
```

```python
sns.boxplot(x='Cloud_Hackathon', y='C++_Score', hue='Placed_Tech', data=dataset);

# 1.Students who participated in cloud hackathons tend to have slightly higher C++ scores.
# 2.Placed students generally have higher C++ scores than non-placed students.
# 3.However, there is significant overlap, indicating that hackathon participation alone is not a strong predictor of placement.
# 4.C++ score appears to have a stronger relationship with placement than hackathon participation.
```

```python
sns.histplot(data=dataset, x='Deep_Learning_Hours', hue='Cloud_Hackathon',kde=True);

# Since both the curves are alomost overlapping then it can be conclude that there is weak relationship between Deep learning Hours and Cloud_Hackathon\
# Also,since this is following Gaussian Distribution , so there is no need of transformation , ML models itself perform well on Normal Distribued data.
```

```python
sns.boxplot(x='Cloud_Hackathon', y='Deep_Learning_Hours', hue='Placed_Tech', data=dataset);


# 1. Placed Students used to study a long in deep learining than non-placed students.
# 2.Even among students who did not participate in hackathons, placed students tend to study more than non-placed students.
# 3. Those who participate in the cloud hackathon, there is not much difference in the no.of hours studied by both placed 
# and non-placed students.
# 4. Students who both participate in hackathons and spend more time studying tend to have better placement outcomes.
```
