# Hyperparameter Tuning (Grid Search CV and Randomized Search CV)

Hyperparameter tuning is the process of finding the best values for a model's hyperparameters to improve its performance.
Parameters are learned automatically during training.

#### Importing required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

#### Importing Dataset

```python
from sklearn import svm,datasets

iris=datasets.load_iris()
iris
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['Flower']=iris.target
df['Flower']=df['Flower'].apply(lambda x: iris.target_names[x])

df.head()
```

#### Splitting Data into Training and Testing Data

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(iris.data,iris.target,test_size=0.3)
```

#### Using SVM Model

```python
model=svm.SVC(kernel='rbf',C=30,gamma='auto')
model.fit(X_train,y_train)
model.score(X_test,y_test)
```

#### Cross Validation Scores for different parameters (Manual)

```python
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(svm.SVC(kernel='linear',C=30,gamma='auto'),iris.data,iris.target,cv=5)
```

```python
cross_val_score(svm.SVC(kernel='rbf',C=30,gamma='auto'),iris.data,iris.target,cv=5)
```

```python
cross_val_score(svm.SVC(kernel='rbf',C=20,gamma='auto'),iris.data,iris.target,cv=5)
```

#### Cross Validation Scores for different parameters (Using for loop)

```python
kernels=['rbf','linear']
C=[1,10,20]
avg_scores={}
for kval in kernels:
    for cval in C:
        cv_scores=cross_val_score(svm.SVC(kernel=kval,C=cval,gamma='auto'),iris.data,iris.target,cv=5)
        avg_scores[kval+'_'+str(cval)]=np.average(cv_scores)
avg_scores
```

## Grid Search CV

#### Cross Validation Scores for different parameters (Scikit-Learn Grid Search CV method): Same work as done above by for loop

```python
from sklearn.model_selection import GridSearchCV

model=svm.SVC(kernel='rbf',C=20,gamma='auto')
              
clf=GridSearchCV(model,{
    'C':[1,10,20],
    'kernel':['rbf','linear']
},cv=5,return_train_score=False)

clf.fit(iris.data,iris.target)
clf.cv_results_
```

```python
df=pd.DataFrame(clf.cv_results_)
```

```python
df.head()
```

```python
df[['param_C','param_kernel','mean_test_score']]
```

```python
dir(clf)
```

```python
clf.best_score_
```

```python
clf.best_params_
```

## Randomized Search CV

#### Cross Validation Scores for different parameters (Scikit-Learn Randomized Search CV Method)

```python
from sklearn.model_selection import RandomizedSearchCV

rs=RandomizedSearchCV(model,{
    'C':[1,10,20],
    'kernel':['rbf','linear']
},cv=5,return_train_score=False,n_iter=2)

rs.fit(iris.data,iris.target)
rs.cv_results_
```

```python
df2=pd.DataFrame(rs.cv_results_)
df2.head()
```

```python
df2[['param_C','param_kernel','mean_test_score']]
```

## Picking Up the Best Model

```python
from sklearn import svm
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

model_params={
    'svm':{
        'model':svm.SVC(gamma='auto'),
        'params':{
            'C':[1,10,20],
            'kernel':['rbf','linear']
        }
    },
    'random_forest':{
        'model':RandomForestClassifier(),
        'params':{
            'n_estimators':[1,5,10]
        }
    },
    'logisitic_regression':{
        'model':LogisticRegression(solver='liblinear',multi_class='auto'),
        'params':{
            'C':[1,5,10]
        }
    }
}
```

```python
scores=[]

for model_name,mp in model_params.items():
    clf=GridSearchCV(mp['model'],mp['params'],cv=5,return_train_score=False)
    clf.fit(iris.data,iris.target)
    scores.append({
        'model':model_name,
        'best_score':clf.best_score_,
        'best_params':clf.best_params_
    })
scores
```

```python
df=pd.DataFrame(scores)
df
```

From the above table ,it can be observed that the SVM model is performing the best.
